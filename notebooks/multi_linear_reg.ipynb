```python
import numpy as np
import matplotlib.pyplot as plt
import math
```

```python
x_train = np.array([[2104, 5, 1, 45],
                   [1416, 3, 2, 40],
                    [852, 2, 1, 35]])


y_train = np.array([460, 232, 178])

print(f"Shape of x_train: {x_train.shape}\n Shape of y_train: {y_train.shape}")
```

```python
print(x_train)
print(f"Type of x_train: {type(x_train)}")
print(y_train)
print(f"Type of y_train: {type(y_train)}")
```

```python
b_init = 785.1811367994083
w_init = np.array([ 0.39133535, 18.75376741, -53.36032453, -26.42131618])
print(f"w_init shape: {w_init.shape}, b_init type: {type(b_init)}")
```

```python
def predict_with_loop(x, w, b):
    res = b

    for i in range(w.shape[0]):
        res += w[i] * x[i]

    return res
```

```python
# get a row from our training data
x_vec = x_train[0,:]
print(f"x_vec shape {x_vec.shape}, x_vec value: {x_vec}")

# make a prediction
f_wb = predict_with_loop(x_vec, w_init, b_init)
print(f"f_wb shape {f_wb.shape}, prediction: {f_wb}")
```

```python
def predict(x, w, b):
    return np.dot(w, x) + b
```

```python
# get a row from our training data
x_vec = x_train[0,:]
print(f"x_vec shape {x_vec.shape}, x_vec value: {x_vec}")

# make a prediction
f_wb = predict(x_vec, w_init, b_init)
print(f"f_wb shape {f_wb.shape}, prediction: {f_wb}")
```

```python
def compute_cost(x, y, w, b):
    total = 0
    m = x.shape[0]
    for i in range(m):
        y_estimate = predict(x[i], w, b)
        diff = (y_estimate - y[i]) ** 2
        total += diff

    result = total / (2 *m)
    return result
```

```python
cost = compute_cost(x_train, y_train, w_init, b_init)
print(f"The cost: {cost}")
```

```python
def compute_gradient(x, y ,w, b):
    m,n = x.shape
    dj_dw = np.zeros(n)
    dj_db = 0

    for i in range(m):
        y_estimate = predict(x[i], w, b)
        diff = y_estimate - y[i]
        dj_db +=  diff
        for j in range(n):
            dj_dw[j] += diff * x[i][j]

    dj_dw /= m
    dj_db /= m

    return dj_db, dj_dw
```

```python
#Compute and display gradient 
tmp_dj_db, tmp_dj_dw = compute_gradient(x_train, y_train, w_init, b_init)
print(f'dj_db at initial w,b: {tmp_dj_db}')
print(f'dj_dw at initial w,b: \n {tmp_dj_dw}')
```

```python
def gradient_descent(x, y, w_in, b_in, learning_rate, num_iters):
    m,n = x.shape
    w = w_in
    b = b_in
    j_history = []

    for i in range(num_iters):
        dj_db, dj_dw = compute_gradient(x, y, w, b)
        w -= learning_rate * dj_dw
        b -= learning_rate * dj_db

        if i < 100000:
            j_history.append(compute_cost(x,y, w, b))

        if i% math.ceil(num_iters / 10) == 0:
            print(f"Iteration {i:4d}: Cost {j_history[-1]:8.2f}   ")

    return w, b, j_history
```

```python
# initialize parameters
initial_w = np.zeros_like(w_init)
initial_b = 0.
# some gradient descent settings
iterations = 1000
alpha = 5.0e-7
# run gradient descent 
w_final, b_final, J_hist = gradient_descent(x_train, y_train, initial_w, initial_b, alpha, iterations)
print(f"b,w found by gradient descent: {b_final:0.2f},{w_final} ")
m,_ = x_train.shape
for i in range(m):
    print(f"prediction: {np.dot(x_train[i], w_final) + b_final:0.2f}, target value: {y_train[i]}")
```
