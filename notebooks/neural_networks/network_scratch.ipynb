```python
import numpy as np
from lab_utils_common import sigmoid
from lab_coffee_utils import load_coffee_data
import tensorflow as tf
```

```python
X, y = load_coffee_data()
(X.shape, y.shape)
```

```python
# normalize
normal_layer = tf.keras.layers.Normalization(axis=-1)
normal_layer.adapt(X)
X_norm = normal_layer(X)
```

```python
g = sigmoid
```

```python
def my_dense(a_in, W, b):
    # A is 1x2 matrix
    # W is 2xunits matrix
    # b is 1xunits
    #a_out should be 1xunits, 
    z = np.matmul(a_in, W) + b
    a_out = g(z)

    return a_out
```

```python
def my_sequential(x, W1, b1, W2, b2):
    a1 = my_dense(x,W1, b1)
    a2 = my_dense(a1, W2, b2)

    return a2
```

```python
W1_tmp = np.array( [[-8.93,  0.29, 12.9 ], [-0.1,  -7.32, 10.81]] )
b1_tmp = np.array( [-9.82, -9.28,  0.96] )
W2_tmp = np.array( [[-31.18], [-27.59], [-32.56]] )
b2_tmp = np.array( [15.41] )
```

```python
def my_predict(X, W1, b1, W2, b2):
    m, _ = X.shape
    y_hat = np.zeros((m,1))
    for i in range(m):
        a_m = X[i]
        y_hat[i] = my_sequential(a_m, W1, b1, W2, b2)
    return y_hat
```

```python
X_tst = np.array([
    [200,13.9],  # postive example
    [200,17]])   # negative example
X_tstn = normal_layer(X_tst)  # remember to normalize
predictions = my_predict(X_tstn, W1_tmp, b1_tmp, W2_tmp, b2_tmp)
```

```python
y_hat = (predictions >= 0.5).astype(int)
print(f"decisions = \n{y_hat}")
```
