```python
import numpy as np
import tensorflow as tf
from lab_coffee_utils import load_coffee_data
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
X, y = load_coffee_data()
print(X.shape, y.shape)
```

```python
normalization_layer = tf.keras.layers.Normalization(axis=-1)
normalization_layer.adapt(X)
X_norm = normalization_layer(X)
```

```python
Xt = np.tile(X_norm,(1000,1))
Yt= np.tile(y,(1000,1))   
```

```python
tf.random.set_seed(1234)  # applied to achieve consistent results
model = Sequential(
    [tf.keras.Input((2,)),
    Dense(units=3, activation="sigmoid", name = "layer1"),
    Dense(units=1, activation="sigmoid", name = "layer2")]
)
```

```python
model.summary()
```

```python
W1, b1 = model.get_layer("layer1").get_weights()
W2, b2 = model.get_layer("layer2").get_weights()
print(f"W1{W1.shape}:\n", W1, f"\nb1{b1.shape}:", b1)
print(f"W2{W2.shape}:\n", W2, f"\nb2{b2.shape}:", b2)
```

```python
model.compile(
    loss = tf.keras.losses.BinaryCrossentropy(),
    optimizer = tf.keras.optimizers.Adam(learning_rate=0.01),
)
```

```python
model.fit(
    Xt,Yt,            
    epochs=10,
)
```

```python
W1, b1 = model.get_layer("layer1").get_weights()
W2, b2 = model.get_layer("layer2").get_weights()
print("W1:\n", W1, "\nb1:", b1)
print("W2:\n", W2, "\nb2:", b2)
```

```python
X_test = np.array([
    [200,13.9],  # positive example
    [200,17]])   # negative example
X_testn = normalization_layer(X_test)
predictions = model.predict(X_testn)
print("predictions = \n", predictions, "\nType: ", type(predictions))
```
